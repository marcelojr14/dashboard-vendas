```python
import numpy as np

pecas_vendidas = np.array([150, 120, 90, 210, 300, 250, 180])
faturamento = pecas_vendidas * 50
dias_de_pico = pecas_vendidas > 200
pecas_acima_200 = pecas_vendidas[dias_de_pico]

print("NÍVEL 1")
print("Peças vendidas:", pecas_vendidas)
print("Faturamento:", faturamento)
print("Dias com mais de 200 peças:", pecas_acima_200)

vendas = np.array([
    [200, 220, np.nan, 250],
    [150, 180, 160, 190],
    [300, 310, 290, 330]
])

total_por_loja = np.nansum(vendas, axis=1)
media_diaria = np.nanmean(vendas)

resultado_meta = np.where(
    vendas >= 200,
    'Meta Atingida',
    'Abaixo'
)

print("\nNÍVEL 2")
print("Vendas:")
print(vendas)
print("Total por loja:", total_por_loja)
print("Média geral:", media_diaria)
print("Resultado das metas:")
print(resultado_meta)

rng = np.random.default_rng(42)

categorias = np.array([
    'Eletrônicos',
    'Roupas',
    'Casa'
])

vendas_clientes = rng.choice(
    categorias,
    size=50
)

categorias_unicas, quantidades = np.unique(
    vendas_clientes,
    return_counts=True
)

faturamento_campanhas = np.array([
    12000,
    45000,
    23000,
    89000,
    31000
])

indice_campea = np.argmax(faturamento_campanhas)

print("\nNÍVEL 3")
print("Categorias das 50 vendas:")
print(vendas_clientes)
print("Categorias:", categorias_unicas)
print("Quantidade por categoria:", quantidades)
print("Índice da campanha campeã:", indice_campea)
print("Faturamento da campanha campeã:", faturamento_campanhas[indice_campea])

dias_mes = np.arange(1, 31)

metas = np.linspace(20000, 30000, 5)

estoque = np.array([10.5, 20.1, 30.9])
estoque_inteiro = estoque.astype(int)

ultimos_5_dias = dias_mes[-5:][::-1]

print("\n\nATIVIDADE 2 - NÍVEL 1")
print("Dias do mês:", dias_mes)
print("Metas:", metas)
print("Estoque inteiro:", estoque_inteiro)
print("Últimos 5 dias invertidos:", ultimos_5_dias)

visitas = np.array([
    1000, 1200, 1100,
    1300, 1400, 1500,
    1600, 1550, 1700,
    1800, 1900, 2000
])

visitas_trimestres = visitas.reshape(4, 3)

primeiro_semestre = np.array([
    [1000, 1200, 1100],
    [1300, 1400, 1500]
])

segundo_semestre = np.array([
    [1600, 1550, 1700],
    [1800, 1900, 2000]
])

ano_completo = np.vstack([
    primeiro_semestre,
    segundo_semestre
])

primeiro_trimestre = ano_completo[0].copy()
ano_achatado = ano_completo.ravel()

print("\nATIVIDADE 2 - NÍVEL 2")
print("Visitas por trimestre:")
print(visitas_trimestres)
print("Ano completo:")
print(ano_completo)
print("Primeiro trimestre:", primeiro_trimestre)
print("Array achatado:", ano_achatado)

pontuacoes = np.array([85, 92, 78, 95, 88])
ranking = np.argsort(pontuacoes)

precos = np.array([100, 200, 300])
precos_coluna = precos[:, np.newaxis]

descontos = np.array([0.10, 0.20, 0.30])
precos_com_desconto = precos_coluna * (1 - descontos)

A = np.array([
    [2, 1],
    [1, -1]
])

b = np.array([
    500,
    100
])

valores = np.linalg.solve(A, b)

print("\nATIVIDADE 2 - NÍVEL 3")
print("Pontuações:", pontuacoes)
print("Índices do ranking:", ranking)
print("Preços como coluna:")
print(precos_coluna)
print("Preços com descontos:")
print(precos_com_desconto)
print("Valor da placa robótica: R$", valores[0])
print("Valor do sensor: R$", valores[1])
```
