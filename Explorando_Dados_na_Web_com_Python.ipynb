```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
from io import StringIO

headers = {
    "User-Agent": "Projeto-Explorando-Dados-Web-Python"
}

url = "https://jsonplaceholder.typicode.com/posts"

resposta = requests.get(
    url,
    params={"userId": 2, "_limit": 5},
    headers=headers,
    timeout=10
)

print("NÍVEL 1")
print("Status:", resposta.status_code)
print("URL final:", resposta.url)
print("Dados:", resposta.json())

ceps = [
    "01001000",
    "20040002",
    "30130010"
]

resultados_ceps = []

for cep in ceps:
    resposta = requests.get(
        f"https://viacep.com.br/ws/{cep}/json/",
        timeout=10
    )
    dados = resposta.json()
    resultados_ceps.append(dados)

df_ceps = pd.DataFrame(resultados_ceps)

print("\nNÍVEL 2")
print("Dados dos CEPs:")
print(df_ceps)

def baixar_arquivo(url, nome_arquivo):
    try:
        resposta = requests.get(url, timeout=10)
        resposta.raise_for_status()

        with open(nome_arquivo, "wb") as arquivo:
            arquivo.write(resposta.content)

        print(f"Arquivo {nome_arquivo} baixado com sucesso.")

    except requests.exceptions.RequestException as erro:
        print(f"Não foi possível realizar o download: {erro}")

baixar_arquivo(
    "https://picsum.photos/400/400",
    "imagem.jpg"
)

robots_url = "https://books.toscrape.com/robots.txt"

resposta_robots = requests.get(
    robots_url,
    headers=headers,
    timeout=10
)

print("\nNÍVEL 3")
print("Robots.txt:")
print(resposta_robots.text)

url_livros = "https://books.toscrape.com/"

resposta = requests.get(
    url_livros,
    headers=headers,
    timeout=10
)

resposta.raise_for_status()

soup = BeautifulSoup(
    resposta.text,
    "html.parser"
)

livros = soup.select("article.product_pod")[:5]

dados_livros = []

for livro in livros:
    titulo = livro.find("h3").find("a")["title"]
    preco = livro.find("p", class_="price_color").text

    dados_livros.append({
        "Titulo": titulo,
        "Preco": preco
    })

df_livros = pd.DataFrame(dados_livros)

print("\nPrimeiros 5 livros:")
print(df_livros)

df_livros.to_csv(
    "livros.csv",
    index=False,
    encoding="utf-8"
)

print("\nArquivo livros.csv criado com sucesso.")

url_wikipedia = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"

resposta_wikipedia = requests.get(
    url_wikipedia,
    headers=headers,
    timeout=10
)

resposta_wikipedia.raise_for_status()

tabelas = pd.read_html(
    StringIO(resposta_wikipedia.text)
)

df_wikipedia = tabelas[0]

print("\nTabela da Wikipedia:")
print(df_wikipedia.head())
```
